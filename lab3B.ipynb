```python
# using class employee of above question create a list of employee and display a list of employee in sorted
# order according to their names.also define fuction to sort list of employee according to their salary in 
# decreasing order.
```

```python
class employee:
    def getdata(self, empid, name, salary):
        self.empid = empid
        self.name = name
        self.salary = salary
        return self

    def showdata(self):
        print("Employee I'd = ", self.empid)
        print("Name = ", self.name)
        print("Salary = ", self.salary)


def sortname(employees):
    return sorted(employees, key=lambda emp: emp.name)


def sortsalary(employees):
    return sorted(employees, key=lambda emp: emp.salary, reverse=True)

emp1=employee()
emp2=employee()
emp3=employee()
employees = [
    emp1.getdata(1, "sonu", 50000),
    emp2.getdata(2, "yash", 90000),
    emp3.getdata(3, "shubham", 200000)
]
print("sorted by name")
sorted_by_name = sortname(employees)
for n in sorted_by_name:
    n.showdata()
print("Sorted by salary")
sorted_by_salary = sortsalary(employees)
for n in sorted_by_salary:
    n.showdata()
```
