```python
class test:
    a=10
    def __init__(self,x,y):
        self.x=x
        test.y=y
    @staticmethod
    def f2(m,n):
        test.d=m
        test.b=n

    @classmethod
    def f3(cls):
        cls.e=100
        cls.f=200

test.g=400
t1=test(10,20)
test.f2(30,30)
test.f3()
print(t1.__dict__)
# print(test.__dict__)
```
