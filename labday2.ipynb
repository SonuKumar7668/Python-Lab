```python
# 1. Write a python script to find number of vowels in each if the string in a given list of strings use map function
def count_vowels(string):
    vowels = "aeiouAEIOU"
    return sum(1 for char in string if char in vowels)

# List of strings
strings = ["hello", "world", "python", "vowels", "example"]

# Using map to apply count_vowels to each string in the list
vowel_counts = list(map(count_vowels, strings))

# Display the result
print(vowel_counts)
```

```python
# 2. Write a python script to find number of digit in each of the element in a given tuple of numbers.Use map function
def count_digits(n):
    return len(str(abs(n)))
t = (1, 122, 3124, 41, 578, 612334)
x = list(map(count_digits,t))
print(x)
```
