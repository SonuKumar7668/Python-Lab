```python
# Write a program using class and object.Assign a instance variable name and object.Assign a instance variable 
# name and salary to employee and static variable employee id and generate unique employee id automatically by 
# incrementing it by 1
```

```python
class Employee:
    # Static variable to keep track of the employee ID
    _employee_id_counter = 1

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        self.employee_id = Employee._employee_id_counter
        Employee._employee_id_counter += 1

    def display_employee_details(self):
        print(f"Employee ID: {self.employee_id}, Name: {self.name}, Salary: {self.salary}")

# Creating instances of Employee
emp1 = Employee("Alice", 50000)
emp2 = Employee("Bob", 60000)
emp3 = Employee("Charlie", 70000)

# Displaying employee details
emp1.display_employee_details()
emp2.display_employee_details()
emp3.display_employee_details()
```
