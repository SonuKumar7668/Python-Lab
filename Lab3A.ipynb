```python
# Define a class employee with instance variable: employee id,name,Salary.
# Define a function to show employee data.
```

```python
class employee:
    def getdata(self,empid,name,salary):
        self.empid=empid
        self.name=name
        self.salary=salary

    def showdata(self):
        print("Employee I'd = ",self.empid)
        print("Name = ",self.name)
        print("Salary = ",self.salary)

emp=employee()
emp.getdata(1,"sonu",50000)
emp.showdata()
```
