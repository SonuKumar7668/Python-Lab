```python
class publication:
    def getdata(self,title,price):
        self.title=title
        self.price=price

class book(publication):
    def getpage(self,count):
        self.pages=count

class cd(book):
    def getcd(self,time):
        self.playminute=time
    def putdata(self):
        print("book Title ->",self.title)
        print("book price",self.price)
        print("book page count ->",self.pages)
        print("cd play time ->",self.playminute)
book1=cd()
book1.getdata("book1",100)
book1.getpage(350)
book1.getcd(500)
book1.putdata()
```

```python
# lab 7
class person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

class student(person):
    def getstudent(self,g,sid):
        self.grade=g
        self.student_id=sid

class teacher(student):
    def getteacher(self,sub,tid):
        self.subject=sub
        self.teacher_id=tid
    def displayinfo(self):
        print(self.name)
        print(self.age)
        print(self.grade)
        print(self.student_id)
        print(self.subject)
        print(self.teacher_id)

ob=teacher("sonu",20)
ob.getstudent("d",5)
ob.getteacher("python",1)
ob.displayinfo()
```
