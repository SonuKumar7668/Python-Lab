```python
'''lab 9 Create a Python program to demonstrate data hiding using a private attribute in a class.
Define a class CounterClass with a private attribute privateCount. 
Implement a method count() to increment and display the private attribute. 
Display the counter value number of times it is called.'''
```

```python
class CounterClass:
    __privateCount=0
    def __init__(self):
        self.__privateCount = 0

    def count(self):
        self.__privateCount += 1
        print(f"Counter value: {self.__privateCount}")

counter = CounterClass()

counter.count()
counter.count()
counter.count()
```
