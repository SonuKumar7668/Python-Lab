```python
class p:
    def m(self):
        print("this is parent class")

class c(p):
    def m1(self):
        print("this is child class")

ob=c()
ob.m1()
ob.m()
```

```python
class p:
    def m(self):
        print("this is parent class")

class c(p):
    def m(self):
        print("this is child class")

bt1=c()
bt2=p()
bt1.m()
bt2.m()
```

```python
class p:
    def m1(self):
        print("this is parent class")

class c(p):
    def m2(self):
        print("this is child class")

class g(c):
    def m3(self):
        print("this is grandchild class")

bt=g()
bt.m1()
bt.m2()
bt.m3()
```

```python
class p:
    def __init__(self):
        print("this is parent constructor")

    def m(self):
        print("this is parent class")

class c(p):
    def __init__(self):
        super().__init__()
        print("this is child constructor")

    def m1(self):
        print("this is child class")

bt=c()
bt.m1()
bt.m()
```

```python
class person:
    def __init__(self,n,a):
        self.name=n
        self.age=a
    def showname(self):
        print(self.name)
    def showage(self):
        print(self.age)

class student(person):
    def rollno(self,r):
        self.roll=r
    def showroll(self):
        print(self.roll)

bt=student("sonu",20)
bt.showname()
bt.showage()
bt.rollno("D05")
bt.showroll()
```
