```python
"""Design a python program demonstrating hybrid inheritance where : Class vehicle has basic attributes like brand and 
model. class car and bike inherit from vehicle and add attributes sapecific to each. Car includes seating capacity. 
bike includes engine type . class electric vehicle inherits from both car and bike to represent electric vehicles. 
it adds attributes like battery capacity. and write a fucntion to print them."""
```

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def print_details(self):
        print(f"Brand: {self.brand}")
        print(f"Model: {self.model}")

class Car(Vehicle):
    def __init__(self, brand, model, seating_capacity):
        Vehicle.__init__(self, brand, model)
        self.seating_capacity = seating_capacity

    def print_details(self):
        Vehicle.print_details(self)
        print(f"Seating Capacity: {self.seating_capacity}")

class Bike(Vehicle):
    def __init__(self, brand, model, engine_type):
        Vehicle.__init__(self, brand, model)
        self.engine_type = engine_type

    def print_details(self):
        Vehicle.print_details(self)
        print(f"Engine Type: {self.engine_type}")

class ElectricVehicle(Car, Bike):
    def __init__(self, brand, model, seating_capacity, engine_type, battery_capacity):
        Car.__init__(self, brand, model, seating_capacity)
        Bike.__init__(self, brand, model, engine_type)
        self.battery_capacity = battery_capacity

    def print_details(self):
        Vehicle.print_details(self)
        print(f"Seating Capacity: {self.seating_capacity}")
        print(f"Engine Type: {self.engine_type}")
        print(f"Battery Capacity: {self.battery_capacity}")
        
ev = ElectricVehicle("Tesla", "Model X", 5, "Electric", "100 kWh")

ev.print_details()
```
